### 1.4 Funciones, Reutilización y Type Hints
Escribir funciones modulares con type hints y docstrings previene bugs en la ingeniería de datos y en los pipelines de ML.

```python
# Demo de Funciones Reutilizables
from typing import List, Dict, Union

def min_max_normalize(data: List[float]) -> List[float]:
    """
    Normaliza una lista numérica al rango [0.0, 1.0].
    
    Fórmula: x_norm = (x - min) / (max - min)
    """
    min_val = min(data)
    max_val = max(data)
    
    if max_val == min_val:
        return [0.0 for _ in data]
        
    return [(x - min_val) / (max_val - min_val) for x in data]

raw_scores = [40.0, 60.0, 80.0, 100.0]
normalized = min_max_normalize(raw_scores)
print(f"Puntajes originales  : {raw_scores}")
print(f"Puntajes normalizados: {normalized}")
```

### 2.1 La Necesidad de Velocidad: Bucles de Python vs. Vectorización
¿Por qué no podemos usar simplemente listas de Python para aprendizaje automático? Ejecutemos un benchmark de rendimiento usando `%timeit`.

```python
# Benchmark de Velocidad - Listas de Python vs. ndarray de NumPy
import time

N = 1_000_000
py_list_a = list(range(N))
py_list_b = list(range(N))

np_arr_a = np.arange(N)
np_arr_b = np.arange(N)

# 1. Bucle Puro de Python
t0 = time.perf_counter()
py_result = [a + b for a, b in zip(py_list_a, py_list_b)]
t_python = time.perf_counter() - t0

# 2. Suma Vectorizada de NumPy
t0 = time.perf_counter()
np_result = np_arr_a + np_arr_b
t_numpy = time.perf_counter() - t0

speedup = t_python / t_numpy
print(f"Tiempo de Bucle Puro de Python : {t_python * 1000:.2f} ms")
print(f"Tiempo Vectorizado de NumPy    : {t_numpy * 1000:.2f} ms")
print(f"NumPy es aproximadamente {speedup:.1f}x más rápido")
```

### 2.4 Broadcasting y Álgebra Matricial (Vista Previa de la Ecuación Normal)

#### ¿Qué es Broadcasting?
Broadcasting le permite a NumPy realizar operaciones aritméticas sobre arreglos de diferentes formas sin copiar datos.
Por ejemplo: centrar datos restando el vector medio de features:
$$\tilde{X} = X - \boldsymbol{\mu}$$
donde $X \in \mathbb{R}^{N \times D}$ y $\boldsymbol{\mu} \in \mathbb{R}^{D}$.

#### Multiplicación de Matrices y Sistemas Lineales
- Multiplicación elemento a elemento: `A * B`
- Producto matricial (producto punto): `A @ B` o `np.dot(A, B)`
- Resolver $A \mathbf{w} = \mathbf{b}$: `np.linalg.solve(A, b)`

En la Regresión Lineal por Mínimos Cuadrados Ordinarios (OLS), los pesos óptimos $\mathbf{w}^*$ se obtienen analíticamente mediante la **Ecuación Normal**:
$$\mathbf{w}^* = (X^T X)^{-1} X^T \mathbf{y} \iff (X^T X) \mathbf{w}^* = X^T \mathbf{y}$$

```python
# Álgebra Matricial y Sistemas Lineales
# Resolvamos un sistema lineal 2D:
# 2*w0 + 1*w1 = 8
# 1*w0 + 3*w1 = 14
A = np.array([[2.0, 1.0],
              [1.0, 3.0]])
b = np.array([8.0, 14.0])

# Resolver para w:
w = np.linalg.solve(A, b)
print(f"Pesos resueltos w: {w}")
print(f"Verificación A @ w: {A @ w} (debería coincidir con b: {b})")

# Norma Euclídea del Vector (norma L2)
v = np.array([3.0, 4.0])
norm_v = np.linalg.norm(v)
print(f"Norma L2 de [3, 4] = {norm_v:.2f}")
```

### 3.3 Selección: `.loc[]` vs. `.iloc[]` (La Trampa Crítica del Indexado)

Una fuente clásica de bugs en pipelines de datos es confundir el indexado **basado en etiqueta** con el **basado en posición**:
- **`.loc[etiqueta_fila, etiqueta_col]`**: Accede por **Etiqueta de Índice** (el nombre del índice o columna).
- **`.iloc[pos_fila, pos_col]`**: Accede por **Posición Entera** (siempre $0, 1, 2, \dots, N-1$).

¡Demostremos qué pasa cuando se eliminan filas y los índices se vuelven no-contiguos!

```python
# Demostración de .loc vs .iloc
# Crear un pequeño dataframe con etiquetas explícitas
df_sample = df.head(5).copy()
df_sample.index = ['a', 'b', 'c', 'd', 'e']

print("DataFrame con etiquetas de índice personalizadas:")
print(df_sample[['sex', 'age', 'fare']])

# Accediendo a la fila 'c'
print("\nAcceso vía .loc['c', 'age'] (Etiqueta)      :", df_sample.loc['c', 'age'])
print("Acceso vía .iloc[2, 1] (Posición 2, Col 1)   :", df_sample.iloc[2, 1])

# ¿Qué pasa si eliminamos la fila 'b'?
df_filtered = df_sample.drop('b')
print("\nDespués de eliminar la fila 'b':")
print(f"Etiquetas de índice presentes: {list(df_filtered.index)}")
print("df_filtered.iloc[1] da la fila:", df_filtered.index[1], "(la posición 1 ahora es la etiqueta 'c')")
```

### 5.1 Matriz de Features ($X$) y Vector Objetivo ($y$)
Scikit-Learn requiere:
- **$X$ (Matriz de Features)**: Un arreglo 2D o DataFrame de forma `(n_samples, n_features)`.
- **$y$ (Vector Objetivo)**: Un arreglo 1D o Series de forma `(n_samples,)` que contiene las etiquetas verdaderas.

```python
# Extracción de Features
# Preparar datos limpios para modelado
model_df = df[['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare']].copy()

# Imputar cualquier valor faltante restante en age con la mediana
model_df['age'] = model_df['age'].fillna(model_df['age'].median())

# Separar X e y
feature_names = ['pclass', 'age', 'sibsp', 'parch', 'fare']
X = model_df[feature_names]
y = model_df['survived']

print(f"Forma de la Matriz de Features X : {X.shape}  (2D)")
print(f"Forma del Vector Objetivo y      : {y.shape}   (1D)")
print("\nPrimeras 3 filas de X:")
display(X.head(3))
```

### 5.2 División Train-Test y Prevención de Fuga de Datos
¿Por qué dividimos los datos? Si evaluamos un modelo sobre los mismos datos con los que fue entrenado, simplemente puede memorizar las entradas (**overfitting**). Debemos probar sobre **datos de test no vistos** para estimar el rendimiento real de generalización.

> [!IMPORTANT]
> **Regla de Fuga de Datos:** Cualquier transformación de preprocesamiento (como el escalado media/desv.est.) debe calcularse **SOLO sobre `X_train`**, y luego aplicarse a `X_test`. ¡Nunca ajustes los scalers sobre el dataset completo!

```python
# División Train/Test y Escalado de Features
# División: 80% Entrenamiento, 20% Testing (estratificado por distribución de clase)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y
)

print(f"Muestras de entrenamiento : {X_train.shape[0]}")
print(f"Muestras de testing       : {X_test.shape[0]}")

# Estandarización de Features: z = (x - mu) / sigma
scaler = StandardScaler()

# ¡Nota: fit_transform en TRAIN, pero solo transform en TEST!
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\nEscalado de features completo. Media del feature 0 de train:", round(X_train_scaled[:, 0].mean(), 4))
print("Desviación estándar del feature 0 de train             :", round(X_train_scaled[:, 0].std(), 4))
```

### 5.3 Entrenamiento y Evaluación del Modelo
Instanciamos un clasificador `LogisticRegression`, lo entrenamos vía `.fit()` y evaluamos sus predicciones con `.predict()`.

```python
# Entrenamiento del Modelo y Predicción
# 1. Instanciar el estimador
clf = LogisticRegression(random_state=42)

# 2. Ajustar el modelo al conjunto de entrenamiento
clf.fit(X_train_scaled, y_train)

# 3. Predecir sobre datos de test no vistos
y_pred = clf.predict(X_test_scaled)

# 4. Calcular Métricas
test_acc = accuracy_score(y_test, y_pred)
print(f"🎯 Accuracy en el Conjunto de Test: {test_acc * 100:.2f}%\n")

print("--- Reporte Detallado de Clasificación ---")
print(classification_report(y_test, y_pred, target_names=["No sobrevivió (0)", "Sobrevivió (1)"]))
```

```python
# Visualización de la Matriz de Confusión
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 4.5))
sns.heatmap(
    cm, 
    annot=True, 
    fmt='d', 
    cmap='RdPu', 
    xticklabels=["Pred: Murió", "Pred: Sobrevivió"],
    yticklabels=["Real: Murió", "Real: Sobrevivió"]
)
plt.title(f"Matriz de Confusión (Accuracy de Test: {test_acc*100:.1f}%)", fontweight='bold')
plt.ylabel("Etiqueta Real")
plt.xlabel("Etiqueta Predicha")
plt.show()
```
